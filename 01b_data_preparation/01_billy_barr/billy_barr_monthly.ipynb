```python
import pandas as pd
import numpy as np
import datetime as dt
import xarray as xr
```

```python

wy22_monthly_rain = [0.93,0,0,0,0,0,0,0.09,1.7,1.84,2.67,1.59]
wy23_monthly_rain = [0.97,0,0,0,0,0,0,0.67,1.44,0.04,1.03,1.24]
normal_monthly_rain = [0.65,0,0,0,0,0,0.06,0.45,1.03,2.22,2.36,2.12]

wy22_monthly_swe = [1.32,1.45,9.77,1.15,2.13,2.99,3.13,1.34,0,0,0,0]
wy23_monthly_swe = [1.35,1.75,5.17,7.15,2.94,9.24,1.82,0.76,0.44,0,0,0]
normal_monthly_swe = [1.83,3.25,3.83,3.89,4.26,3.94,3.22,1.99,0.24,0,0,0.40]

wy22_monthly_total = np.array(wy22_monthly_rain) + np.array(wy22_monthly_swe)
wy23_monthly_total = np.array(wy23_monthly_rain) + np.array(wy23_monthly_swe)
normal_monthly_total = np.array(normal_monthly_rain) + np.array(normal_monthly_swe)

months = ['Oct','Nov','Dec','Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep']

df = pd.DataFrame({
    'Month': months,
    'WY22_Rain_in': wy22_monthly_rain,
    'WY23_Rain_in': wy23_monthly_rain,
    'Normal_Rain_in': normal_monthly_rain,
    'WY22_SWE_in': wy22_monthly_swe,
    'WY23_SWE_in': wy23_monthly_swe,
    'Normal_SWE_in': normal_monthly_swe,
    'WY22_Total_in': wy22_monthly_total,
    'WY23_Total_in': wy23_monthly_total,
    'Normal_Total_in': normal_monthly_total
}).set_index('Month')
```

```python
wy22_accum_norm = df.loc[['Nov','Dec','Jan','Feb','Mar']]['WY22_Total_in'].sum()/df.loc[['Nov','Dec','Jan','Feb','Mar']]['Normal_Total_in'].sum()
wy23_accum_norm = df.loc[['Nov','Dec','Jan','Feb','Mar']]['WY23_Total_in'].sum()/df.loc[['Nov','Dec','Jan','Feb','Mar']]['Normal_Total_in'].sum()

wy22_melt_norm = df.loc[['Apr','May','Jun']]['WY22_Total_in'].sum()/df.loc[['Apr','May','Jun']]['Normal_Total_in'].sum()
wy23_melt_norm = df.loc[['Apr','May','Jun']]['WY23_Total_in'].sum()/df.loc[['Apr','May','Jun']]['Normal_Total_in'].sum()


wy22_monsoon_norm = df.loc[['Jul','Aug','Sep']]['WY22_Total_in'].sum()/df.loc[['Jul','Aug','Sep']]['Normal_Total_in'].sum()
wy23_monsoon_norm = df.loc[['Jul','Aug','Sep']]['WY23_Total_in'].sum()/df.loc[['Jul','Aug','Sep']]['Normal_Total_in'].sum()

wy22_total_norm = df['WY22_Total_in'].sum()/df['Normal_Total_in'].sum()
wy23_total_norm = df['WY23_Total_in'].sum()/df['Normal_Total_in'].sum()

print("WY22 accum Season Total / Normal:", wy22_accum_norm)
print("WY23 accum Season Total / Normal:", wy23_accum_norm)
print("WY22 Melt Season Total / Normal:", wy22_melt_norm)
print("WY23 Melt Season Total / Normal:", wy23_melt_norm)
print("WY22 Monsoon Season Total / Normal:", wy22_monsoon_norm)
print("WY23 Monsoon Season Total / Normal:", wy23_monsoon_norm)
print("WY22 Annual Total / Normal:", wy22_total_norm)
print("WY23 Annual Total / Normal:", wy23_total_norm)
```
